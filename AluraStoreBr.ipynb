### Importação dos dados



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja.head(100)

```

#1. Análise do faturamento


```python

#loja0
faturamento_liquido = loja['Preço'] - loja['Frete']
faturamento_total = faturamento_liquido.sum()
print(f'Faturamento total da Loja é: R${faturamento_total:.2f}\n')

#loja2
faturamento_liquido2 = loja2['Preço'] - loja2['Frete']
faturamento_total2 = faturamento_liquido2.sum()
print(f'Faturamento total da Loja é: R${faturamento_total2:.2f}\n')

#loja3
faturamento_liquido3 = loja3['Preço'] - loja3['Frete']
faturamento_total3 = faturamento_liquido3.sum()
print(f'Faturamento total da Loja é: R${faturamento_total3:.2f}\n')

#loja4
faturamento_liquido4 = loja4['Preço'] - loja4['Frete']
faturamento_total4 = faturamento_liquido4.sum()
print(f'Faturamento total da Loja é: R${faturamento_total4:.2f}\n')

lista = [faturamento_total, faturamento_total2, faturamento_total3, faturamento_total4]
nomes_lojas = ['Loja 0', 'Loja 2', 'Loja 3', 'Loja 4']
maior_valor = max(lista)
indice_maior = lista.index(maior_valor)
loja_maior_faturamento = nomes_lojas[indice_maior]

print(f"A loja com o maior faturamento total é: {loja_maior_faturamento} com R${maior_valor:.2f}")
```

# 2. Vendas por Categoria


```python
def analisar_loja_por_categoria(df, nome_loja):
    df['Faturamento_Liquido'] = df['Preço'] - df['Frete']

    resultado = df.groupby('Categoria do Produto').agg(
        Faturamento_Liquido=('Faturamento_Liquido', 'sum'),
        Numero_Vendas=('Categoria do Produto', 'count')
    ).reset_index()

    resultado['Loja'] = nome_loja  # adiciona a identificação da loja
    resultado['Numero_Vendas'] = resultado['Numero_Vendas'].astype(int)
    return round(resultado,2)

# Analisando cada loja separadamente
analise_loja = analisar_loja_por_categoria(loja, 'Loja 0')
analise_loja2 = analisar_loja_por_categoria(loja2, 'Loja 2')
analise_loja3 = analisar_loja_por_categoria(loja3, 'Loja 3')
analise_loja4 = analisar_loja_por_categoria(loja4, 'Loja 4')

# (Opcional) Concatenar tudo em um só DataFrame, caso queira comparar
analise_completa = pd.concat([analise_loja, analise_loja2, analise_loja3, analise_loja4])

# Ordenar por categoria e depois pelo faturamento (do maior para o menor)
analise_completa = analise_completa.sort_values(
    by=['Categoria do Produto', 'Faturamento_Liquido'],
    ascending=[True, False]
).reset_index(drop=True)

# Mostrar resultado final
print(analise_completa)
```
